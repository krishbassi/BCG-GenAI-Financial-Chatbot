```python
def simple_chatbot(user_query):
    user_query = user_query.lower()

    if user_query == "what is microsoft's total revenue in 2025?":
        return "Microsoft's total revenue in 2025 was $281,724 million."
    elif user_query == "how did tesla's net income change in 2025?":
        return "Tesla's net income decreased by approximately 46.1% in 2025."
    elif user_query == "what was apple's operating cash flow in 2025?":
        return "Apple's 2025 operating cash flow was $111,482 million."
    else:
        return "Sorry, I didn't understand that.Please ask one of the predefined financial questions."

user_query = input("Ask a financial question: ")
print(simple_chatbot(user_query))
```

## Chatbot Documentation

This prototype is a rule-based financial chatbot developed in Python. It processes a user's question by removing surrounding whitespace and converting the text to lowercase, making query matching more consistent. The `simple_chatbot()` function then uses `if`, `elif`, and `else` statements to compare the question against predefined financial queries and return a corresponding response based on the financial analysis completed in Task 1.

The chatbot currently responds to three predefined questions covering Microsoft revenue, Tesla net income growth, and Apple operating cash flow. Testing confirmed that all three recognised queries return the expected response and that an unsupported query triggers a fallback response.

The main limitation is that the chatbot relies on exact predefined query matching and hard-coded responses. It does not use natural language processing or machine learning, so differently worded questions with the same meaning may not be recognised.

A future version could retrieve financial information dynamically from the underlying dataset and use natural language processing to interpret a wider range of financial questions.
